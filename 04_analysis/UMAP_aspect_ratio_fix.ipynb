This notebook is entirely for aethetic purposes so that we have high quality UMAP images for our figures.

```python
import scanpy as sc
import matplotlib.pyplot as plt
```

```python
new_ad = sc.read_h5ad('Transciptomic_labels_and_morphology_labels_full.h5ad')
```

```python
fig, ax = plt.subplots(figsize=(5,4.5))
sc.pl.umap(new_ad, color=["sub_mic"],size=50, ax=ax)
fig.savefig('umap_clusters.png', dpi=300, bbox_inches='tight')#,save='_clusters.png')
```

```python
fig, ax = plt.subplots(figsize=(5,4.5))
sc.pl.umap(new_ad, color=["Apoe"],size=50,cmap='hot', ax=ax)
fig.savefig('umap_apoe.png', dpi=300, bbox_inches='tight')
```

```python
fig, ax = plt.subplots(figsize=(5,4.5))
sc.pl.umap(new_ad, color=["Tmem119"],cmap='hot',size=50, ax=ax)
fig.savefig('umap_tmem119.png', dpi=300, bbox_inches='tight')
```

```python
fig, ax = plt.subplots(figsize=(5,4.5))
sc.pl.umap(new_ad, color=["Hif1a"],cmap='hot',size=50, ax=ax)
fig.savefig('umap_hif1a.png', dpi=300, bbox_inches='tight')
```

```python
fig, ax = plt.subplots(figsize=(5,4.5))
sc.pl.umap(new_ad[new_ad.obs.Brain_Region != 'Unlabeled'], color=["Brain_Region"],size=50, ax=ax)
fig.savefig('umap_brain_region.png', dpi=300, bbox_inches='tight')
```

```python
fig, ax = plt.subplots(figsize=(5,5))
sc.pl.embedding(new_ad, basis = 'X_umap_shapes', color=["ordered_morph"],size=50, ax=ax)
fig.savefig('morphology_clusters.png', dpi=300, bbox_inches='tight')
```

```python
fig, ax = plt.subplots(figsize=(5,5))
sc.pl.embedding(new_ad, basis = 'X_umap_shapes', color=["sub_mic"],size=50, ax=ax)
fig.savefig('morphology_transcriptomic_clusters.png', dpi=300, bbox_inches='tight')
```
